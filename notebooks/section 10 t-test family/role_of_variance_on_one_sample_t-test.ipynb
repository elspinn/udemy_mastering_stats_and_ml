```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
from matplotlib.pyplot import figure
```

```python
# For array of variance
variances = np.linspace(0.1, 3, 100)
# Generate sample observations
n = 20
create_observations = lambda variance: variance * np.random.randn(n) + 0.5
observations = np.array([create_observations(var) for var in variances])
# Compute one sample t-test, t & p
pop_mean = 0
stds = np.array([np.std(obs) for obs in observations])
t_tests = np.array([stats.ttest_1samp(obs, pop_mean) for obs in observations])
t = t_tests[..., 0]
p = t_tests[..., 1]
```

```python
fig, ax = plt.subplot_mosaic([['t'],['p'],['p_t']], figsize=[6, 12])
fig.set_layout_engine('constrained')

ax['p'].plot(variances, t, 'ko', markersize=2)
ax['p'].set_xlabel('Variance')
ax['p'].set_ylabel('T')

ax['t'].plot(variances, p, 'ko', markersize=2)
ax['t'].set_xlabel('Variance')
ax['t'].set_ylabel('P')

ax['p_t'].plot(np.absolute(t), p, 'ko', markersize=1)
ax['p_t'].set_title('One Sample T-Test')
ax['p_t'].set_xlabel('abs(T)')
ax['p_t'].set_ylabel('P')

```
